## 2) Build the nitromethane molecule and write it to file
We'll create a single nitromethane geometry using RDKit, optimize it, and write an XYZ/PDB that PACKMOL or ASE can use.

```
from rdkit import Chem
from rdkit.Chem import AllChem
from ase import Atoms
import numpy as np

smiles = 'C[N+](=O)[O-]'  # nitromethane SMILES used earlier
mol = Chem.MolFromSmiles(smiles)
mol = Chem.AddHs(mol)
AllChem.EmbedMolecule(mol, randomSeed=42)
AllChem.MMFFOptimizeMolecule(mol)
conf = mol.GetConformer()
symbols = [a.GetSymbol() for a in mol.GetAtoms()]
positions = conf.GetPositions()
ase_mol = Atoms(symbols=symbols, positions=positions)
# Write xyz and pdb for use with PACKMOL or converters
ase_mol.write('nm.xyz')
ase_mol.write('nm.pdb')
print('Wrote nm.xyz and nm.pdb with', len(ase_mol), 'atoms')
```

## 3) Create a condensed-phase box (PACKMOL or ASE fallback)
We'll compute a target box size from density (1.97 g/cm^3) and create a PACKMOL input. If PACKMOL is available, the notebook will run it; otherwise a simple ASE-based random packing fallback is used (not ideal but useful for demonstrations).

```
import math, os
from ase.io import read, write
from ase.build import molecule

def compute_box_side(n_mol=16, mol_weight=61.04, density_g_cm3=1.97):
    # mol_weight in g/mol, density in g/cm^3
    mass_g = n_mol * mol_weight / 6.022e23
    vol_cm3 = mass_g / density_g_cm3
    side_A = (vol_cm3 * 1e24) ** (1/3)
    return side_A

n_mol = 16
side = compute_box_side(n_mol)
print('Target box side (A) for', n_mol, 'molecules at 1.97 g/cm^3:', side)

# prepare PACKMOL input file
packmol_in = f'''tolerance 2.0
filetype xyz
output nm_box.xyz

structure nm.xyz
  number {n_mol}
  inside box 0. 0. 0. {side:.6f} {side:.6f} {side:.6f}
end structure
'''
open('packmol_nm.inp','w').write(packmol_in)
print('Wrote packmol_nm.inp — box side', side)

# Try to run packmol if binary available
if os.path.exists('/usr/bin/packmol') or os.system('which packmol >/dev/null 2>&1') == 0:
    print('PACKMOL found — running...')
    os.system('packmol < packmol_nm.inp')
    print('PACKMOL finished — wrote nm_box.xyz')
else:
    print('PACKMOL not found — creating a simple random packing fallback (not for production)')
    # simple random packing: replicate molecule positions randomly inside box avoiding overlaps
    from ase import Atoms
    base = read('nm.xyz')
    positions = []
    symbols = []
    rng = np.random.default_rng(12345)
    attempts = 0
    while len(positions) < n_mol * len(base) and attempts < 1000000:
        mol_index = len(positions) // len(base)
        # place a new molecule center randomly
        shift = rng.uniform(0, side, size=3)
        for p,s in zip(base.get_positions(), base.get_chemical_symbols()):
            positions.append(p + shift)
            symbols.append(s)
        attempts += 1
    atoms = Atoms(symbols=symbols, positions=positions)
    atoms.set_cell([side,side,side])
    atoms.center()
    write('nm_box.xyz', atoms)
    print('Wrote nm_box.xyz using ASE fallback with', len(atoms)//len(base), 'molecules (approx)')
```

## 4) Convert PACKMOL / XYZ to LAMMPS data file
You can convert with `molecule` tools (topotools in VMD), `pymatgen`, `MDTraj`, or ASE. The following cell writes a LAMMPS data file using ASE's write(lammps-data). You will need to ensure atom types and masses match the ReaxFF parameter ordering (C H N O).

```
from ase.io import read, write
atoms = read('nm_box.xyz')
# ASE's lammps-data writer requires some metadata about masses/types; we keep a minimal file and user should edit types to match ffield.reax ordering
write('nm_box.data', atoms, format='lammps-data')
print('Wrote nm_box.data — edit atom types/masses in the file as needed to match your parameter file')
```

## 5) Write a LAMMPS ReaxFF input template
This template expects you to upload `ffield.reax` (the ReaxFF CHNO parameter file) into the Colab filesystem or into the repo. Replace file names and adjust settings as needed for your ReaxFF parameter set.

```
lammps_template = '''# LAMMPS input for ReaxFF nitromethane (template)
units real
atom_style charge
boundary p p p

read_data nm_box.data

pair_style reax/c NULL
pair_coeff * * ffield.reax C H N O

neighbor 2.0 bin
neigh_modify every 1 delay 0 check yes

fix qeq all qeq/reax 1 0.0 10.0 1e-6 reax/c

timestep 0.25
thermo 500
thermo_style custom step temp etotal press

# species and bonds output (if available in your LAMMPS build)
# fix sp all reaxff/species 100 species.dat
# fix bd all reaxff/bonds 100 bonds.dat

# coordinate dump
dump traj all custom 1000 traj.lammpstrj id type x y z
dump_modify traj sort id

# initialize velocities at target temperature (adjust seed)
velocity all create 3000.0 12345 mom yes rot yes dist gaussian

fix nvt all nvt temp 3000.0 3000.0 100.0
run 800000

unfix qeq
write_dump all custom final_dump.lammpstrj id type x y z
'''
open('in.reax_nm','w').write(lammps_template)
print('Wrote in.reax_nm — edit it to match your parameter file and desired run lengths')
```

## 7) Analysis: parsing species/bonds and trajectories
The following analysis functions parse common ReaxFF outputs and reproduce the plots from the demo. Use ReaxFF outputs if available (fix reaxff/species, fix reaxff/bonds or dumped bond-orders). If you only have coordinate dumps, the code falls back to distance-cutoff bond tests (use cutoffs from the paper's supporting information).

```
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from MDAnalysis import Universe

def read_species_file(path='species.dat'):
    # user may need to adjust parsing depending on LAMMPS build format
    try:
        df = pd.read_csv(path, delim_whitespace=True, comment='#', header=None)
        return df
    except Exception as e:
        print('Could not read species file:', e)
        return None

def read_lammps_dump(topology='nm_box.data', traj='traj.lammpstrj'):
    try:
        u = Universe(topology, traj)
        return u
    except Exception as e:
        print('Error reading dump with MDAnalysis:', e)
        return None

def detect_first_bond_break(u, atom_pairs, cutoff=1.8):
    # atom_pairs: list of tuples of LAMMPS atom IDs (1-based) or atom selection strings
    times = []
    for a,b in atom_pairs:
        broken_at = None
        for ts in u.trajectory:
            pa = u.atoms[a-1].position
            pb = u.atoms[b-1].position
            d = np.linalg.norm(pa-pb)
            if d > cutoff:
                broken_at = ts.time / 1000.0 if ts.time is not None else None
                break
        times.append(broken_at)
    return times

print('Analysis utilities defined. Use read_species_file(), read_lammps_dump() and plotting cells below.')
```

## 8) Demo / mock mode (quick end-to-end check)
If you cannot run ReaxFF in Colab, run the demo mode which uses the synthetic energy/species data and the original plotting code so you can verify all plotting/analysis cells run correctly.

```
# Synthetic plotting code (adapted from your original demo)
import numpy as np
import matplotlib.pyplot as plt
plt.style.use('seaborn-v0_8-whitegrid')
def synthetic_energy(T, t_max=100, n=500):
    t = np.linspace(0, t_max, n)
    if T == 3000:
        ind = 5.0; drop = 80
    elif T == 2500:
        ind = 15.0; drop = 60
    else:
        ind = 40.0; drop = 40
    E = np.zeros_like(t)
    mask = t >= ind
    E[mask] = -drop * (1 - np.exp(-(t[mask]-ind)/8))
    E += np.random.normal(0, 1.5, n)
    return t, E

fig = plt.figure(figsize=(12,8))
ax1 = plt.subplot(2,2,1)
for T,color in zip([3000,2500,2000], ['C0','C1','C2']):
    t,E = synthetic_energy(T)
    ax1.plot(t,E,label=f'T={T}K', color=color)
    ax1.axvline({3000:0.5,2500:1.5,2000:5.0}[T], color=color, ls='--', alpha=0.6)
ax1.set_xlabel('Time (ps)'); ax1.set_ylabel('Potential energy (kcal/mol per NM)'); ax1.legend()

ax2 = plt.subplot(2,2,2)
t = np.linspace(0,100,300)
for T,P0,dP in zip([3000,2500,2000],[22,20,18],[7,5,5]):
    P = P0 + dP*(1-np.exp(-np.maximum(t-5,0)/15)) + np.random.normal(0,0.4,len(t))
    ax2.plot(t,P,label=f'T={T}K')
ax2.set_xlabel('Time (ps)'); ax2.set_ylabel('Pressure (GPa)'); ax2.legend()

ax3 = plt.subplot(2,2,3)
# small mock Table 1 bars
configs = ['crystal3000','liquid13000','liquid23000']
x = np.arange(3)
ax3.bar(x-0.2,[0.1,0.05,0.3],0.2,label='C-H')
ax3.bar(x,[2.8,0.5,0.7],0.2,label='C-N')
ax3.set_yscale('log'); ax3.set_xticks(x); ax3.set_xticklabels(configs,rotation=45)
ax3.legend(fontsize=8)

ax4 = plt.subplot(2,2,4)
t = np.linspace(0,150,400)
nm_pop = np.exp(-t/25)
h2o = 0.8*(1-np.exp(-(t-8)/20))*(t>5)
ax4.plot(t,nm_pop,'g-',label='Nitromethane')
ax4.plot(t,h2o,'b-',label='H2O')
ax4.set_xlim(0,150); ax4.legend()

plt.tight_layout()
plt.show()
```
